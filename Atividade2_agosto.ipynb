```python
#questao 4

horas = int(input("digite o numero de horas passadas: "))
valor = 0

if horas <= 2:
    valor = valor + 8

elif horas > 2 and horas <= 5:
    hora_adicional = horas - 2
    valor = (valor + 8) + hora_adicional * 2

else:
    hora_adicional = horas - 5
    valor = (valor + 17) + hora_adicional * 2

print("o valor a ser pago foi de", valor ,"reais")
print("o numero de horas passadas foi de", horas ,"horas")
```

```python
#questao 11

numero = int(input("digite um numero maior que 1: "))

divisores = 0

for i in range(2, numero):
    if numero % i == 0:
        divisores = divisores + 1

if divisores == 0:
    print("primo")
else:
    print("composto")
```

```python
#questao 13

A = int(input("digite um numero A (positivo): "))
B = int(input("digite um numero B (positivo, maior que A): "))

contador_P = 0

for numero in range(A, B + 1):
    divisores = 0
    for i in range(2, numero):
        if numero % i == 0:
            divisores = divisores + 1

    if divisores == 0 and numero > 1:
        print(numero)
        contador_P = contador_P + 1

print("Quantidade de primos encontrados:", contador_P)
```

```python
#questao 14

n_secreto = 37
tentativas = 0

while tentativas < 5:
    N_D = int(input("digite um numero: "))

    if N_D > n_secreto:
        print("tente um numero menor")
        tentativas += 1

    elif N_D < n_secreto:
        print("tente um numero maior")
        tentativas += 1

    else:
        print("voce acertou")
        tentativas += 1
        break

if N_D == n_secreto:
    print("você acertou e foram", tentativas, "tentativas")
else:
    print("você não acertou. O número era", n_secreto)
```

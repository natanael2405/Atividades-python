```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

n_digitado = int(input("digite um numero (0 para parar): "))

while n_digitado > 0 or n_digitado < 0:
    quantidade = quantidade + 1
    soma = soma + n_digitado

    if n_digitado > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

    n_digitado = int(input("digite um numero (0 para parar): "))

print("quantidade de numeros digitados:", quantidade)
print("quantidade de positivos:", positivos)
print("quantidade de negativos:", negativos)
print("soma de todos os numeros:", soma)
```

```python
# questao 5

listaN = []

for i in range(8):
    n = int(input("digite um numero: "))
    listaN.append(n)

maior = listaN[0]
menor = listaN[0]
posMaior = 0
posMenor = 0
posicao = 0

for n in listaN:
    if n > maior:
        maior = n
        posMaior = posicao
    if n < menor:
        menor = n
        posMenor = posicao
    posicao = posicao + 1

print("maior valor:", maior)
print("Apareceu na posicao", posMaior)
print("menor valor:", menor)
print("Apareceu na posicao", posMenor)
```

```python
#questao 11

contagem = {}

for i in range(8):
    palavra = input("Digite uma palavra: ")

    if palavra in contagem:
        contagem[palavra] = contagem[palavra] + 1
    else:
        contagem[palavra] = 1

for palavra in contagem:
    quantidade = contagem[palavra]
    print(palavra, "apareceu", quantidade, "vez(es)")
```

```python
#questao 12

dic1 = {}

dic1['arroz'] = 15
dic1['feijão'] = 20
dic1['macarrao'] = 15
dic1['carne'] = 30

pesquisa = input("pesquise o produto: ")
venda = int(input("digite a quantidade à ser vendida: "))

if pesquisa in dic1:

    if venda <= dic1[pesquisa]:

        if pesquisa == 'arroz':
            dic1['arroz'] = dic1['arroz'] - venda

        elif pesquisa == 'feijão':
            dic1['feijão'] = dic1['feijão'] - venda

        elif pesquisa == 'macarrao':
            dic1['macarrao'] = dic1['macarrao'] - venda

        elif pesquisa == 'carne':
            dic1['carne'] = dic1['carne'] - venda

    else:
        print("estoque insuficiente")

else:
    print("produto nao cadastrado")

print(dic1)
```

```python
#questao 13

turma = []

for i in range(5):
    nome = input("Nome do aluno: ")
    nota1 = float(input("Primeira nota: "))
    nota2 = float(input("Segunda nota: "))
    media = (nota1 + nota2) / 2

    aluno = {
        "nome": nome,
        "notas": [nota1, nota2],
        "media": media
    }

    turma.append(aluno)

print("---Resultado Final---")
for aluno in turma:
    print(aluno["nome"], ": média", aluno["media"])
```

```python
#questao 15

catalogo = [
    {"codigo": 1, "nome": "arroz", "quantidade": 25, "preco": 30},
    {"codigo": 2, "nome": "feijao", "quantidade": 40, "preco": 15},
    {"codigo": 3, "nome": "cafe", "quantidade": 30, "preco": 12},
    {"codigo": 4, "nome": "açucar", "quantidade": 18, "preco": 14},
    {"codigo": 5, "nome": "macarrão", "quantidade": 13, "preco": 10},
]


for item in catalogo:
    print(item)

VTE = 0
itens_C = 0
for item in catalogo:
    VTE = VTE + item["quantidade"] * item["preco"]
    itens_C = itens_C + 1

print("Valor total:", VTE)

maior = catalogo[0]
for item in catalogo:
    if item["quantidade"] * item["preco"] > maior["quantidade"] * maior["preco"]:
        maior = item

print("Item de maior valor:", maior["nome"])


resumo = (itens_C, "itens em estoque" ,VTE,"reais em estoque")
print("Resumo:", resumo)
```
